```python
import requests
from bs4 import BeautifulSoup as bs
github_user = input("Enter gihub user")
url = "https://github.com/"+github_user
r = requests.get(url)
soup = bs(r.content,'html.parser')
profile_image = soup.find("img",{"alt":"Avatar"})
if profile_image is not None:
    profile_image = profile_image.get("src")
print(profile_image)
```

```python
import requests
from bs4 import BeautifulSoup as bs

github_user = input("Enter GitHub username: ").strip()
url = f"https://github.com/{github_user}"

headers = {
    "User-Agent": "Mozilla/5.0"  # Avoid blocking
}

r = requests.get(url, headers=headers)

if r.status_code == 200:
    soup = bs(r.text, 'html.parser')

    # Look for the img tag inside the user profile avatar container
    profile_image_tag = soup.find("img", class_="avatar avatar-user width-full border color-bg-default")

    if profile_image_tag:
        profile_image_url = profile_image_tag.get("src")
        print("Profile image URL:", profile_image_url)
    else:
        print("Profile image not found.")
else:
    print("User not found or request failed with status code:", r.status_code)
```

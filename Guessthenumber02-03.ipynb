```python
import random

def guess(x):
  random_number = random.randint(1,x)
  guess = 0
  while(guess != random_number):
    guess = int(input(f"choose number between 1 and {x}"))
    if guess < random_number:
      print("sory guess again too low")
    elif guess > random_number:
      print("soory guess too high")

    print(f"congrat your guess is {random_number}")

def computer_guess(x):
  low=1
  high=x
  feedback = ''
  while feedback != 'c':
      if low!= high:
        guess = random.randint(low,high)
      else:
        guess = low
      feedback = input(f"is {guess} too high ,too low or corrected one")
      if feedback == 'h':
        high = guess - 1
      elif feedback == 'l':
        low = guess + 1
      print(f"congrats you guesss {guess}")

guess(10)
computer_guess(20)
```
